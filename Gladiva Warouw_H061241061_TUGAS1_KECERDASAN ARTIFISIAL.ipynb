## Prompt 1 — Interpretasi ERT & Seismic Refraction Tomography (PyGIMLi)

- **role:** Kamu adalah seorang geofisikawan eksplorasi dengan pengalaman lebih dari 10 tahun dalam pemodelan geolistrik (ERT) dan seismik refraksi, terbiasa bekerja dengan Python dan library PyGIMLi untuk inversi data bawah permukaan dangkal, serta memahami cara membandingkan hasil model sintetik dengan studi kasus pada jurnal geofisika terapan.
- **task:** Bangun dan jelaskan alur kerja pemodelan sintetik lapisan tanggul (embankment) menggunakan PyGIMLi, menggabungkan hasil inversi ERT dan tomografi seismik refraksi menjadi satu interpretasi geologi yang koheren.
- **context:** Model mengacu pada studi Penta de Peppo et al. (2024, GJI) tentang deteksi rembesan (seepage) pada tanggul menggunakan kombinasi metode geolistrik dan seismik; tujuan akhirnya adalah menghasilkan figur pembanding gaya publikasi ilmiah.
- **input:** Parameter geometri tanggul (tinggi, lebar puncak, kemiringan lereng), resistivitas dan kecepatan gelombang-P tiap lapisan sintetik, serta skema elektroda/geofon yang digunakan.
- **constraints:** Gunakan hanya library PyGIMLi (pygimli.physics.ert dan pygimli.physics.traveltime); jangan mengubah asumsi fisis dasar (Archie's law untuk resistivitas, hukum Snell untuk refraksi) tanpa disebutkan secara eksplisit; sertakan seed acak jika ada noise sintetik agar hasil dapat direproduksi.
- **output:** Dua artifact — (1) skrip Python lengkap dengan komentar tiap tahap (mesh, forward modeling, inversi), (2) figur 2×2 yang membandingkan model true vs. hasil inversi ERT dan seismik.
- **quality_criteria:** Hasil inversi harus konvergen (chi-square mendekati 1), figur harus memilig skala warna dan colorbar yang konsisten dengan model asli, dan kode harus berjalan tanpa error dari awal hingga akhir.
- **verification (optional):** Jika chi-square jauh dari 1 atau model geologi tidak masuk akal secara fisis, minta AI melakukan sensitivity check pada parameter regularisasi (lambda) sebelum melaporkan hasil akhir.

## Prompt 2 — Pemrosesan & Interpretasi Anomali Magnetik (Oasis Montaj)

- **role:** Kamu adalah seorang ahli geofisika eksplorasi mineral yang berspesialisasi dalam survei magnetik udara/darat, sangat menguasai perangkat lunak Oasis Montaj (modul MAGMAP) serta teknik filter turunan medan potensial.
- **task:** Susun prosedur pengolahan data anomali magnetik total (TMI) dari tahap gridding hingga penerapan filter RTP, RTE, dan turunan (FHD, TDR, TDX) untuk mengidentifikasi batas struktur geologi bawah permukaan.
- **context:** Data berasal dari survei magnetik regional pada zona lintang rendah (Indonesia), sehingga koreksi inklinasi-deklinasi IGRF dan proyeksi UTM zona 50S menjadi krusial sebelum interpretasi kualitatif.
- **input:** Data grid TMI mentah (format .grd atau .xyz), nilai IGRF (inklinasi, deklinasi, intensitas total) pada lokasi dan waktu survei, serta parameter grid (cell size, metode interpolasi Minimum Curvature).
- **constraints:** Urutan pemrosesan harus mengikuti kaidah baku (koreksi IGRF → gridding Minimum Curvature → RTP → filter turunan); jangan melompati tahap RTP sebelum menerapkan filter turunan karena akan mendistorsi interpretasi struktur pada lintang rendah.
- **output:** Panduan langkah demi langkah (dapat berupa tabel prosedur) lengkap dengan parameter yang direkomendasikan untuk tiap filter, serta deskripsi bagaimana membaca pola anomali hasil TDR/TDX sebagai indikasi kontak litologi atau sesar.
- **quality_criteria:** Prosedur harus konsisten dengan praktik standar geofisika eksplorasi (bukan asumsi bebas), setiap filter dijelaskan tujuan fisisnya (bukan hanya cara klik pada software), dan hasil akhir dapat ditelusuri kembali ke data mentah.
- **verification (optional):** Jika pola anomali RTP tampak tidak simetris terhadap sumber yang diketahui, minta AI mengevaluasi ulang parameter inklinasi/deklinasi IGRF yang digunakan sebelum lanjut ke filter turunan.

## Prompt 3 — Pemilihan Teori Gelombang Nonlinear & Prediksi Gelombang Pecah (Hidrodinamika Pesisir)

- **role:** Kamu adalah seorang oseanografer fisik yang fokus pada dinamika gelombang pesisir, memahami secara mendalam teori gelombang linear (Airy), Stokes orde tinggi, cnoidal, dan solitary wave, serta kriteria pemilihan teori berdasarkan parameter kedalaman relatif dan kecuraman gelombang.
- **task:** Tentukan teori gelombang yang paling tepat digunakan untuk suatu kondisi perairan pesisir tertentu, lalu prediksikan lokasi dan tipe gelombang pecah (spilling, plunging, surging, atau collapsing) berdasarkan kemiringan dasar dan parameter gelombang datang.
- **context:** Kasus diterapkan pada perairan dangkal tropis dengan variasi kedalaman dan kemiringan pantai yang berbeda-beda, sesuai pembahasan Bab 4 (Gelombang Nonlinear, Pecah, dan Pemilihan Teori) pada mata kuliah Hidrodinamika Pesisir.
- **input:** Tinggi gelombang (H), periode (T), kedalaman perairan (d), dan kemiringan dasar pantai (m/tan β) pada lokasi yang ditinjau.
- **constraints:** Gunakan diagram/kriteria baku (misalnya Ursell number atau diagram Le Méhauté) untuk memilih teori gelombang, bukan tebakan; klasifikasi tipe gelombang pecah harus didasarkan pada parameter Iribarren (surf similarity parameter), bukan deskripsi kualitatif semata.
- **output:** Penjelasan naratif berisi (1) nilai parameter tak berdimensi yang dihitung (Ursell number, Iribarren number), (2) teori gelombang yang direkomendasikan beserta alasannya, dan (3) prediksi tipe gelombang pecah dengan justifikasi numerik.
- **quality_criteria:** Setiap kesimpulan harus didukung oleh perhitungan angka (bukan asumsi kualitatif), rentang berlaku tiap teori disebutkan secara eksplisit, dan hasil konsisten dengan diagram klasifikasi standar di literatur hidrodinamika pesisir.
- **verification (optional):** Jika nilai parameter berada tepat di batas antar-kategori (misalnya di perbatasan cnoidal dan Stokes), minta AI menjelaskan implikasi ketidakpastian tersebut terhadap keandalan prediksi gelombang pecah.

## Prompt 4 — Analisis Arus & Angin untuk Keselamatan Jalur Pelayaran (Meteorologi Maritim)

- **role:** Kamu adalah seorang meteorolog maritim yang berpengalaman menganalisis data reanalysis angin dan arus permukaan laut (format NetCDF) untuk mendukung perencanaan rute pelayaran yang aman dan efisien.
- **task:** Analisis pola arus dan angin permukaan pada tanggal tertentu di jalur pelayaran antar-kota, lalu berikan rekomendasi jalur pelayaran yang meminimalkan risiko akibat arus/angin yang berlawanan dengan arah pelayaran.
- **context:** Studi kasus jalur pelayaran regional Asia Tenggara (misalnya rute pelayaran domestik-internasional pendek) menggunakan data angin hourly dan arus harian tahun berjalan, diolah dengan Python (xarray/netCDF4) atau MATLAB (m_map, cmocean).
- **input:** Data angin (komponen u,v) dan arus (komponen u,v) pada tanggal yang ditentukan, koordinat titik awal-tujuan pelayaran, serta batas wilayah peta yang ditinjau.
- **constraints:** Konversi arah dan kecepatan angin/arus harus mengikuti konvensi oseanografi/meteorologi yang benar (arah "from" untuk angin, arah "towards" untuk arus); peta harus mencantumkan skala vektor dan proyeksi yang jelas agar tidak menyesatkan interpretasi.
- **output:** Peta vektor arus dan angin pada wilayah dan tanggal yang diminta, disertai jalur pelayaran yang digambarkan pada peta, serta interpretasi tertulis singkat mengenai segmen jalur yang berisiko (misal arus berlawanan arah > ambang tertentu).
- **quality_criteria:** Peta harus terbaca jelas (legenda, arah panah konsisten dengan data), interpretasi harus merujuk langsung ke nilai kecepatan/arah pada peta (bukan generalisasi), dan rekomendasi jalur harus logis secara oseanografis.
- **verification (optional):** Jika rekomendasi jalur melewati area dengan data kosong/gap, minta AI menandai keterbatasan tersebut dan menyarankan sumber data pelengkap sebelum rekomendasi difinalisasi.

## Prompt 5 — Interpretasi Anomali Gravitasi untuk Identifikasi Struktur Cekungan Bawah Permukaan

- **role:** Kamu adalah seorang geofisikawan yang ahli dalam metode gravitasi untuk eksplorasi cekungan sedimen dan struktur geologi regional, memahami tahapan reduksi data gravitasi (koreksi udara bebas, Bouguer, terrain) hingga pemodelan forward 2D/3D.
- **task:** Olah data anomali gravitasi Bouguer lengkap menjadi peta anomali residual dan regional (melalui filtering/upward continuation), kemudian identifikasi keberadaan serta estimasi kedalaman struktur cekungan sedimen di bawahnya.
- **context:** Data diambil dari wilayah dengan topografi bervariasi (dataran-perbukitan) sehingga koreksi terrain menjadi signifikan; tujuan akhir adalah mendukung studi potensi cekungan hidrokarbon atau air tanah dalam skala regional.
- **input:** Data anomali Bouguer lengkap (grid atau titik pengukuran dengan koordinat), model densitas rata-rata batuan sekitar, dan peta topografi/DEM wilayah kajian untuk koreksi terrain.
- **constraints:** Pemisahan anomali regional-residual harus menggunakan metode yang disebutkan secara eksplisit (misalnya upward continuation atau polynomial trend surface fitting), bukan pemisahan visual sembarangan; estimasi kedalaman harus menggunakan pendekatan kuantitatif (misalnya metode Nettleton, second vertical derivative, atau forward modeling 2D) bukan hanya interpretasi kualitatif peta warna.
- **output:** Peta anomali Bouguer, anomali regional, dan anomali residual (tiga peta terpisah atau subplot), serta estimasi kedalaman dan bentuk geometri cekungan dalam bentuk penampang (cross-section) 2D.
- **quality_criteria:** Metode pemisahan regional-residual dan asumsi densitas harus dinyatakan secara eksplisit dan dapat dipertanggungjawabkan secara fisis, hasil estimasi kedalaman harus disertai rentang ketidakpastian, dan penampang geometri harus konsisten dengan pola anomali residual yang dihasilkan.
- **verification (optional):** Jika hasil estimasi kedalaman berbeda signifikan antar-metode yang dicoba, minta AI membandingkan asumsi masing-masing metode dan menjelaskan metode mana yang paling sesuai dengan konteks geologi wilayah kajian.
